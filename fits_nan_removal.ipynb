```python
from astropy.io import fits
import numpy as np
from tqdm import tqdm
```

```python
path_image_1 = "/home/wesley/Desktop/JWST_REU/2025_data/"
```

```python
hdul_1 = fits.open((path_image_1 + 'jwst-2025-f444w-f470n.fits'))
hdul_2 = fits.open((path_image_1 + 'jwst-2025-f212n.fits'))
```

```python
hdul_1.info()
```

```python
# finds the x and y pixel values for all the nan data in the image

nan_y = np.where(np.isnan(hdul_1[1].data))[0]
nan_x = np.where(np.isnan(hdul_1[1].data))[1]
```

```python
for z in tqdm(range(len(nan_x))):
    """
    This takes in the location of all the nan pixels and will find the value of the 8 pixels surrounding it 
    The mean of the 8 pixels is taken as long as the at least 4 pixels have a real number value
    This preforms a check on each pixel to make sure that the pixel is within the bounds of the data and makes it a nan if not
    It then changes the value of the nan pixel to the mean and saves it to a new fits file
    """
    cx, cy = nan_x[z], nan_y[z]

    p1 = hdul_1[1].data[cy+1, cx-1] if cx-1 >= 0 and cy+1 < hdul_1[1].data.shape[0] else np.nan
    p2 = hdul_1[1].data[cy,   cx-1] if cx-1 >= 0 else np.nan
    p3 = hdul_1[1].data[cy-1, cx-1] if cx-1 >= 0 and cy-1 >= 0 else np.nan

    p4 = hdul_1[1].data[cy+1, cx  ] if cy+1 < hdul_1[1].data.shape[0] else np.nan
    p5 = hdul_1[1].data[cy-1, cx  ] if cy-1 >= 0 else np.nan
    
    p6 = hdul_1[1].data[cy+1, cx+1] if cy+1 < hdul_1[1].data.shape[0] and cx+1 < hdul_1[1].data.shape[1] else np.nan
    p7 = hdul_1[1].data[cy,   cx+1] if cx+1 < hdul_1[1].data.shape[1] else np.nan
    p8 = hdul_1[1].data[cy-1, cx+1] if cy-1 >= 0 and cx+1 < hdul_1[1].data.shape[1] else np.nan
    
    values = [p1, p2, p3, p4, p5, p6, p7, p8]
    if sum(np.isnan(x) for x in values) > 4:
        continue
    
    new_value = np.nanmean(values)
    hdul_1[1].data[nan_y[z], nan_x[z]] = new_value

# hdul_1.writeto(path_image_1 +'nan-rm-jwst-2025-f444w-f470n.fits', overwrite=True)
```
